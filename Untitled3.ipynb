```python
"""
ATE evaluator (RMSE + MAE) for 2-D trajectories.

Folder layout
-------------
a058_3_gt.npy          # ground-truth
a058_3_umgloc.npy      # UMGLoc prediction
...
<idx>_rnin.npy         # RNIN outputs, where <idx> is 0…N-1
"""

import pathlib, re, sys
import numpy as np
from collections import defaultdict

ROOT = pathlib.Path("/output/")     # change if needed
GT_SUFFIX   = "_gt"
UML_SUFFIX  = "_umgloc"
RNI_SUFFIX  = "_rnin"

def compute_ate(pred, gt):
    """Return (RMSE, MAE) after clipping/padding to GT length."""
    n = len(gt)
    if len(pred) >= n:
        pred = pred[:n]
    else:
        pad = np.repeat(pred[-1][None, :], n - len(pred), axis=0)
        pred = np.vstack([pred, pad])

    err = np.linalg.norm(pred - gt, axis=1)
    return np.sqrt(np.mean(err**2)), np.mean(err)

def load_xy(path):
    """Load .npy and keep first two columns."""
    return np.load(path)[:, :2]

def main():
    # ---------------- collect files ----------------
    gt_files   = sorted(ROOT.glob(f"*{GT_SUFFIX}.npy"))
    uml_files  = {f.stem.replace(UML_SUFFIX, ""): f for f in ROOT.glob(f"*{UML_SUFFIX}.npy")}
    rni_files  = sorted(ROOT.glob(f"*{RNI_SUFFIX}.npy"),
                        key=lambda p: int(re.match(r"(\d+)", p.stem).group(1)))

    if not gt_files:
        sys.exit("No *_gt.npy files found.")

    if len(rni_files) < len(gt_files):
        print("⚠️  Fewer RNIN files than ground-truth – some sequences won’t have RNIN ATE.")

    # --------------- evaluate -----------------------
    overall = defaultdict(list)

    for idx, gt_path in enumerate(gt_files):
        seq = gt_path.stem.replace(GT_SUFFIX, "")    # e.g., "a058_3"
        gt_xy = load_xy(gt_path)

        print(f"\nSequence: {seq:<15}  len={len(gt_xy)}")

        # ---------- UMGLoc ----------
        if seq in uml_files:
            uml_xy = load_xy(uml_files[seq])
            rmse, mae = compute_ate(uml_xy, gt_xy)
            overall["umgloc"].append(rmse)
            print(f"  UMGLoc   RMSE={rmse:7.3f} m   MAE={mae:7.3f} m")
        else:
            print("  UMGLoc   (file missing)")

        # ---------- RNIN (by order) ----------
        if idx < len(rni_files):
            rni_xy = load_xy(rni_files[idx])
            rmse, mae = compute_ate(rni_xy, gt_xy)
            overall["rnin"].append(rmse)
            print(f"  RNIN     RMSE={rmse:7.3f} m   MAE={mae:7.3f} m   (from {rni_files[idx].name})")
        else:
            print("  RNIN     (no RNIN file left)")

    # --------------- aggregate ----------------------
    print("\n--- Aggregate RMSE over sequences ---")
    for tag in ["umgloc", "rnin"]:
        if overall[tag]:
            print(f"  {tag:<7}: mean RMSE = {np.mean(overall[tag]):7.3f} m  "
                  f"over {len(overall[tag])} sequences")

if __name__ == "__main__":
    main()
```
