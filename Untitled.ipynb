```python
def bresenham_line(x0, y0, x1, y1):
    """Generate the points on the line from (x0, y0) to (x1, y1) using Bresenham's algorithm."""
    points = []
    dx = abs(x1 - x0)
    dy = abs(y1 - y0)
    sx = 1 if x0 < x1 else -1
    sy = 1 if y0 < y1 else -1
    err = dx - dy
    
    while True:
        points.append((x0, y0))
        if x0 == x1 and y0 == y1:
            break
        e2 = 2 * err
        if e2 > -dy:
            err -= dy
            x0 += sx
        if e2 < dx:
            err += dx
            y0 += sy
    return points

# Example usage
x0, y0 = trajectory_discrete[0]
x1, y1 = trajectory_discrete[-1]
line_points = bresenham_line(x0, y0, x1, y1)

# Plot the metric map
plt.imshow(metric_map, cmap='gray')

# Overlay the Bresenham line on the map
for point in line_points:
    plt.plot(point[1], point[0], 'ro')  # note: y, x order in imshow

plt.title("Line between Start and End Points Mapped onto Metric Map")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import interp1d

# Example metric map and continuous trajectory
metric_map = create_metric_map(size=100)
trajectory_continuous = np.array([[1.5, 1.2], [2.1, 2.3], [3.6, 4.7], [5.8, 6.5], [7.2, 7.9]])

# Assume the map represents a 10m x 10m area, so each cell represents 0.1m
cell_size = 0.1

# Sub-grid factor: how many finer cells per original cell
sub_grid_factor = 5  # e.g., each 0.1m cell is subdivided into 5x5 finer cells

# Function to interpolate the trajectory with higher resolution
def interpolate_trajectory(trajectory, num_points=200):
    t = np.arange(trajectory.shape[0])
    t_new = np.linspace(t.min(), t.max(), num_points)

    x_interpolator = interp1d(t, trajectory[:, 0], kind='linear')
    y_interpolator = interp1d(t, trajectory[:, 1], kind='linear')

    x_new = x_interpolator(t_new)
    y_new = y_interpolator(t_new)

    return np.vstack((x_new, y_new)).T

# Function to apply moving average smoothing
def smooth_trajectory(trajectory, window_size=5):
    smoothed = np.copy(trajectory)
    for i in range(window_size, len(trajectory) - window_size):
        smoothed[i] = np.mean(trajectory[i - window_size:i + window_size + 1], axis=0)
    return smoothed

# Interpolate the trajectory
interpolated_trajectory = interpolate_trajectory(trajectory_continuous, num_points=500)

# Apply smoothing
smoothed_trajectory = smooth_trajectory(interpolated_trajectory, window_size=10)

# Convert smoothed continuous trajectory to sub-grid indices
sub_cell_size = cell_size / sub_grid_factor
trajectory_subgrid = np.floor(smoothed_trajectory / sub_cell_size).astype(int)

# Map sub-grid indices back to original grid for visualization
trajectory_discrete = trajectory_subgrid // sub_grid_factor

# Plot the metric map
plt.imshow(metric_map, cmap='gray')

# Overlay the sub-grid interpolated trajectory on the map
for point in trajectory_discrete:
    plt.plot(point[1], point[0], 'ro')  # note: y, x order in imshow

plt.title("Sub-Grid Interpolated Trajectory Mapped onto Metric Map")
plt.show()
```
