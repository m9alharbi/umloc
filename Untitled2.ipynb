```python
#!/usr/bin/env python3
"""
Plot predicted & ground-truth trajectories (with optional occupancy map).

Usage
-----
python plot_trajectories.py \
    --model1 my_model.npy \
    --model2 methodA.npy \
    --model3 methodB.npy \
    --gt      gt.npy \
    --map_pgm map.pgm \
    --map_yaml map.yaml \
    --out     comparison.png
"""

import argparse
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import yaml
from PIL import Image

# --------------------------- helpers ---------------------------------
def load_yaml(yaml_path):
    """Return resolution (m/pixel) and origin (x,y) in map frame."""
    with open(yaml_path, "r") as f:
        data = yaml.safe_load(f)
    res = data["resolution"]                       # metres per pixel
    origin = data["origin"][:2]                    # (x, y) of map bottom-left
    return res, origin

def load_map(map_pgm, map_yaml):
    """Load pgm as grayscale array + extent in world coordinates."""
    img = Image.open(map_pgm)
    img = np.array(img)[::-1, :]                   # flip vertically (ROS → Matplotlib)
    res, origin = load_yaml(map_yaml)
    h, w = img.shape
    extent = [
        origin[0], origin[0] + w * res,            # x_min, x_max
        origin[1], origin[1] + h * res,            # y_min, y_max
    ]
    return img, extent

def add_trajectory(ax, arr, label, color, zorder):
    """arr: (N,2) => (x,y)"""
    ax.plot(arr[:, 0], arr[:, 1],
            label=label, linewidth=1.6, color=color, zorder=zorder)

# ------------------------ main script --------------------------------
def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--model1", required=True, help=".npy file for Model 1")
    parser.add_argument("--model2", required=True, help=".npy file for Model 2")
    parser.add_argument("--model3", required=True, help=".npy file for Model 3")
    parser.add_argument("--gt",      required=True, help=".npy file for ground truth")
    parser.add_argument("--map_pgm", help=".pgm occupancy grid (optional)")
    parser.add_argument("--map_yaml", help="corresponding YAML (needed with --map_pgm)")
    parser.add_argument("--out", default="trajectory_comparison.png",
                        help="output figure path")
    parser.add_argument("--dpi", type=int, default=400,
                        help="figure DPI (use ≥ 300 for papers)")
    args = parser.parse_args()

    # ---------- load trajectories ----------
    traj1 = np.load(args.model1)   # shape: (N, 2)
    traj2 = np.load(args.model2)
    traj3 = np.load(args.model3)
    gt    = np.load(args.gt)

    # ---------- plot ----------
    plt.style.use("ggplot")
    fig, ax = plt.subplots(figsize=(6, 6))

    # map (if provided)
    if args.map_pgm:
        if not args.map_yaml:
            parser.error("--map_yaml is required when using --map_pgm")
        img, extent = load_map(args.map_pgm, args.map_yaml)
        ax.imshow(img, cmap="gray", extent=extent, origin="lower", zorder=0)

    # trajectories
    add_trajectory(ax, traj1, "Our Model",           "#1f77b4", 3)
    add_trajectory(ax, traj2, "Method A (SOTA 1)",   "#d62728", 4)
    add_trajectory(ax, traj3, "Method B (SOTA 2)",   "#2ca02c", 5)
    add_trajectory(ax, gt,    "Ground Truth",        "#000000", 6)

    # aesthetics
    ax.set_aspect("equal", adjustable="box")
    ax.set_xlabel("X [m]")
    ax.set_ylabel("Y [m]")
    ax.set_title("Trajectory Comparison")
    ax.legend(loc="upper right", fontsize="small", frameon=True)
    ax.grid(False)
    plt.tight_layout()

    # save
    out_path = Path(args.out)
    out_path.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(out_path, dpi=args.dpi, bbox_inches="tight")
    print(f"Figure saved to: {out_path.resolve()}")


if __name__ == "__main__":
    main()
```
